<h2> Data Creation </h2>

_Use this space to create train/test data_

```python
from sklearn.model_selection import train_test_split
```

```python
# @@@ 8
feature_columns = [
    'AVG_READING_8_SCORE',
    'AVG_READING_4_SCORE',
    'INSTRUCTION_EXPENDITURE_PER_STUDENT',
    'TOTAL_REVENUE_PER_STUDENT'
]
target_column = 'AVG_MATH_8_SCORE'

# Keep complete cases so scikit-learn receives no missing predictor values.
model_data = df[feature_columns + [target_column]].replace([np.inf, -np.inf], np.nan).dropna()
X = model_data[feature_columns]
y = model_data[target_column]
```

```python
# @@@ 9 
# ill keep 30% of the observations for the final test with unbiased data.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)
```
